```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Store the ordered list of selected ports
voyage_stops = []

# Port selection dropdown
stop_dropdown = widgets.Dropdown(
    options=port_options,
    description="Select Port:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

# Buttons
add_stop_button = widgets.Button(
    description="Add Stop",
    button_style="success",
    icon="plus"
)

remove_stop_button = widgets.Button(
    description="Remove Last Stop",
    button_style="warning",
    icon="minus"
)

clear_stops_button = widgets.Button(
    description="Clear All",
    button_style="danger",
    icon="trash"
)

# Output area
itinerary_output = widgets.Output()

# Display current itinerary
def display_itinerary():

    with itinerary_output:
        clear_output(wait=True)

        print("VOYAGE PLANNER — ITINERARY")
        print("--------------------------")

        if not voyage_stops:
            print("No ports selected.")
            return

        for index, port_id in enumerate(voyage_stops, start=1):

            port = active_ports[
                active_ports["port_id"] == port_id
            ].iloc[0]

            print(f"{index}. {port['port_name']}")

# Add stop
def add_stop_clicked(button):

    selected_port = stop_dropdown.value

    if voyage_stops and voyage_stops[-1] == selected_port:
        with itinerary_output:
            clear_output(wait=True)
            print("Error: Consecutive stops cannot be the same port.")
        return

    voyage_stops.append(selected_port)

    display_itinerary()

# Remove last stop
def remove_stop_clicked(button):

    if voyage_stops:
        voyage_stops.pop()

    display_itinerary()

# Clear itinerary
def clear_stops_clicked(button):

    voyage_stops.clear()

    display_itinerary()

# Connect buttons
add_stop_button.on_click(add_stop_clicked)
remove_stop_button.on_click(remove_stop_clicked)
clear_stops_button.on_click(clear_stops_clicked)

# Display interface
display(
    widgets.HTML("<h3>Multi-Leg Voyage Planner</h3>"),
    stop_dropdown,
    widgets.HBox([
        add_stop_button,
        remove_stop_button,
        clear_stops_button
    ]),
    itinerary_output
)

display_itinerary()
```

```python
import searoute as sr
import pandas as pd

# Ensure at least two ports are selected
if len(voyage_stops) < 2:
    print("Error: Please select at least two ports.")

else:
    # Store individual leg results
    leg_results = []

    # Calculate each voyage leg
    for i in range(len(voyage_stops) - 1):

        origin_id = voyage_stops[i]
        destination_id = voyage_stops[i + 1]

        # Retrieve port information
        origin = active_ports[
            active_ports["port_id"] == origin_id
        ].iloc[0]

        destination = active_ports[
            active_ports["port_id"] == destination_id
        ].iloc[0]

        # Convert coordinates to searoute format
        origin_coords = [
            float(origin["approach_longitude"]),
            float(origin["approach_latitude"])
        ]

        destination_coords = [
            float(destination["approach_longitude"]),
            float(destination["approach_latitude"])
        ]

        # Calculate maritime route
        route = sr.searoute(
            origin_coords,
            destination_coords,
            units="naut"
        )

        # Extract distance
        distance = route.properties["length"]

        # Store leg information
        leg_results.append({
            "Leg": i + 1,
            "Origin": origin["port_name"],
            "Destination": destination["port_name"],
            "Distance (NM)": distance,
            "Coordinates": route.geometry["coordinates"]
        })

    # Calculate total maritime distance
    total_distance = sum(
        leg["Distance (NM)"] for leg in leg_results
    )

    # Retrieve vessel speed
    vessel_speed = speed_input.value

    # Calculate estimated sailing duration
    total_hours = total_distance / vessel_speed
    total_days = total_hours / 24

    # Display results
    print("VOYAGE PLANNER — MULTI-LEG RESULTS")
    print("----------------------------------")

    # Create results table without geometry
    results_df = pd.DataFrame([
        {
            "Leg": leg["Leg"],
            "Origin": leg["Origin"],
            "Destination": leg["Destination"],
            "Distance (NM)": round(leg["Distance (NM)"], 2)
        }
        for leg in leg_results
    ])

    display(results_df)

    print(f"\nTotal maritime distance: {total_distance:,.2f} NM")
    print(f"Vessel speed: {vessel_speed:.1f} knots")
    print(f"Estimated sailing time: {total_hours:.2f} hours")
    print(f"Estimated sailing duration: {total_days:.2f} days")
```

```python
# ============================================================
# MARITIME VOYAGE PLANNER
# Project 6: Maritime Distance and Voyage Calculation Service
#
# Consolidated Jupyter Notebook Prototype
#
# Features:
# 1. Port-to-Port Voyage Planning
# 2. Multi-Leg Voyage Planning
# 3. Maritime Distance Calculation
# 4. Estimated Sailing Duration
# 5. Passage Restrictions
# 6. Alternative Route Comparison
# 7. Interactive Maritime Maps
# 8. Input Validation
#
# Routing Engine: searoute (prototype)
# UI Framework: ipywidgets
# Map Framework: Folium
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import folium
import searoute as sr
import ipywidgets as widgets

from pathlib import Path
from IPython.display import display, clear_output


# ============================================================
# 2. LOAD PORT DATASET
# ============================================================

DATASET_PATH = Path("data/voyage_planner_datasets.xlsx")

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH.resolve()}"
    )

ports_df = pd.read_excel(
    DATASET_PATH,
    sheet_name="Port Reference"
)

active_ports = ports_df[
    ports_df["status"].astype(str).str.lower() == "active"
].copy()

active_ports = active_ports[
    [
        "port_id",
        "port_name",
        "country",
        "latitude",
        "longitude",
        "approach_latitude",
        "approach_longitude"
    ]
]

active_ports = active_ports.sort_values(
    "port_name"
).reset_index(drop=True)

port_lookup = active_ports.set_index("port_id")

port_options = [
    (
        f"{row['port_name']} ({row['country']})",
        row["port_id"]
    )
    for _, row in active_ports.iterrows()
]

print(f"Successfully loaded {len(active_ports)} active ports.")


# ============================================================
# 3. PASSAGE RESTRICTIONS
# ============================================================

PASSAGE_OPTIONS = [
    ("Automatic — Default Route", []),
    ("Avoid Suez Canal", ["suez"]),
    ("Avoid Panama Canal", ["panama"]),
    (
        "Avoid Suez and Panama Canals",
        ["suez", "panama"]
    ),
    (
        "Avoid Suez, Panama and Northwest Passage",
        ["suez", "panama", "northwest"]
    )
]


# ============================================================
# 4. HELPER FUNCTIONS
# ============================================================

def get_port(port_id):
    """Retrieve port information from the dataset."""

    if port_id not in port_lookup.index:
        raise ValueError(f"Unknown port ID: {port_id}")

    return port_lookup.loc[port_id]


def get_coordinates(port_id):
    """Return coordinates in searoute format [lon, lat]."""

    port = get_port(port_id)

    return [
        float(port["approach_longitude"]),
        float(port["approach_latitude"])
    ]


def calculate_route(origin_id, destination_id, restrictions=None):
    """Calculate a maritime route using searoute."""

    if origin_id == destination_id:
        raise ValueError(
            "Origin and destination cannot be the same."
        )

    if restrictions is None:
        restrictions = []

    origin_coords = get_coordinates(origin_id)
    destination_coords = get_coordinates(destination_id)

    route = sr.searoute(
        origin_coords,
        destination_coords,
        units="naut",
        restrictions=restrictions
    )

    distance = float(route.properties["length"])
    coordinates = route.geometry["coordinates"]

    if not coordinates or distance <= 0:
        raise ValueError("No valid maritime route was returned.")

    return {
        "origin_id": origin_id,
        "destination_id": destination_id,
        "origin": get_port(origin_id)["port_name"],
        "destination": get_port(destination_id)["port_name"],
        "distance_nm": distance,
        "coordinates": coordinates,
        "restrictions": list(restrictions)
    }


def calculate_duration(distance_nm, speed_knots):
    """Calculate estimated sailing time."""

    if speed_knots <= 0:
        raise ValueError(
            "Vessel speed must be greater than zero."
        )

    hours = distance_nm / speed_knots
    days = hours / 24

    return {
        "hours": hours,
        "days": days
    }


def create_map(routes, port_ids):
    """Create an interactive Folium map."""

    voyage_map = folium.Map(
        location=[20, 40],
        zoom_start=2,
        tiles="OpenStreetMap"
    )

    colors = [
        "blue",
        "green",
        "purple",
        "orange",
        "red",
        "darkblue"
    ]

    all_coordinates = []

    for index, route in enumerate(routes):

        coordinates = [
            [lat, lon]
            for lon, lat in route["coordinates"]
        ]

        all_coordinates.extend(coordinates)

        color = colors[index % len(colors)]

        folium.PolyLine(
            locations=coordinates,
            color=color,
            weight=4,
            opacity=0.85,
            tooltip=(
                f"{route['origin']} → "
                f"{route['destination']}: "
                f"{route['distance_nm']:,.2f} NM"
            )
        ).add_to(voyage_map)

    for index, port_id in enumerate(port_ids):

        port = get_port(port_id)

        coordinates = [
            float(port["approach_latitude"]),
            float(port["approach_longitude"])
        ]

        if index == 0:
            marker_color = "green"
        elif index == len(port_ids) - 1:
            marker_color = "red"
        else:
            marker_color = "blue"

        folium.Marker(
            location=coordinates,
            popup=(
                f"Stop {index + 1}: "
                f"{port['port_name']}"
            ),
            tooltip=port["port_name"],
            icon=folium.Icon(
                color=marker_color,
                icon="info-sign"
            )
        ).add_to(voyage_map)

    if all_coordinates:
        voyage_map.fit_bounds(all_coordinates)

    return voyage_map


# ============================================================
# 5. SHARED APPLICATION STATE
# ============================================================

latest_result = None
voyage_stops = []


# ============================================================
# 6. SHARED UI COMPONENTS
# ============================================================

speed_input = widgets.BoundedFloatText(
    value=15.0,
    min=1.0,
    max=40.0,
    step=0.5,
    description="Vessel Speed (knots):",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="350px")
)

restriction_dropdown = widgets.Dropdown(
    options=PASSAGE_OPTIONS,
    description="Passage Restrictions:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="500px")
)


# ============================================================
# 7. PORT-TO-PORT VOYAGE PLANNER
# ============================================================

origin_dropdown = widgets.Dropdown(
    options=port_options,
    description="Origin:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

destination_dropdown = widgets.Dropdown(
    options=port_options,
    description="Destination:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

port_route_button = widgets.Button(
    description="Calculate Route",
    button_style="primary",
    icon="ship"
)

port_output = widgets.Output()


def calculate_port_route_clicked(button):

    global latest_result

    with port_output:

        clear_output(wait=True)

        latest_result = None

        try:
            origin_id = origin_dropdown.value
            destination_id = destination_dropdown.value

            speed = float(speed_input.value)
            restrictions = list(restriction_dropdown.value)

            route = calculate_route(
                origin_id,
                destination_id,
                restrictions
            )

            duration = calculate_duration(
                route["distance_nm"],
                speed
            )

            latest_result = {
                "voyage_type": "port_to_port",
                "ports": [origin_id, destination_id],
                "speed_knots": speed,
                "restrictions": restrictions,
                "total_distance_nm": route["distance_nm"],
                "duration_hours": duration["hours"],
                "duration_days": duration["days"],
                "legs": [
                    {
                        "leg_number": 1,
                        **route
                    }
                ]
            }

            print("PORT-TO-PORT VOYAGE SUMMARY")
            print("----------------------------")

            print(f"Origin: {route['origin']}")
            print(f"Destination: {route['destination']}")

            print(
                "Restrictions:",
                ", ".join(restrictions) or "None"
            )

            print(
                f"Maritime Distance: "
                f"{route['distance_nm']:,.2f} NM"
            )

            print(f"Vessel Speed: {speed:.1f} knots")

            print(
                f"Estimated Sailing Time: "
                f"{duration['hours']:.2f} hours"
            )

            print(
                f"Estimated Sailing Duration: "
                f"{duration['days']:.2f} days"
            )

            voyage_map = create_map(
                [route],
                [origin_id, destination_id]
            )

            display(voyage_map)

        except Exception as error:
            print("Route calculation failed.")
            print("Error:", error)


port_route_button.on_click(calculate_port_route_clicked)


# ============================================================
# 8. MULTI-LEG VOYAGE PLANNER
# ============================================================

stop_dropdown = widgets.Dropdown(
    options=port_options,
    description="Select Port:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

add_stop_button = widgets.Button(
    description="Add Stop",
    button_style="success",
    icon="plus"
)

remove_stop_button = widgets.Button(
    description="Remove Last Stop",
    button_style="warning",
    icon="minus"
)

clear_stops_button = widgets.Button(
    description="Clear All",
    button_style="danger",
    icon="trash"
)

calculate_multi_button = widgets.Button(
    description="Calculate Multi-Leg Voyage",
    button_style="primary",
    icon="ship"
)

itinerary_output = widgets.Output()
multi_output = widgets.Output()


def display_itinerary():

    with itinerary_output:

        clear_output(wait=True)

        print("VOYAGE ITINERARY")
        print("----------------")

        if not voyage_stops:
            print("No ports selected.")
            return

        for index, port_id in enumerate(voyage_stops, 1):

            port = get_port(port_id)

            print(f"{index}. {port['port_name']}")


def add_stop_clicked(button):

    global latest_result

    selected_port = stop_dropdown.value

    if voyage_stops and voyage_stops[-1] == selected_port:

        with itinerary_output:
            clear_output(wait=True)
            print(
                "Error: Consecutive stops "
                "cannot be the same port."
            )

        return

    voyage_stops.append(selected_port)
    latest_result = None

    display_itinerary()


def remove_stop_clicked(button):

    global latest_result

    if voyage_stops:
        voyage_stops.pop()

    latest_result = None
    display_itinerary()


def clear_stops_clicked(button):

    global latest_result

    voyage_stops.clear()
    latest_result = None

    display_itinerary()


add_stop_button.on_click(add_stop_clicked)
remove_stop_button.on_click(remove_stop_clicked)
clear_stops_button.on_click(clear_stops_clicked)


def calculate_multi_clicked(button):

    global latest_result

    with multi_output:

        clear_output(wait=True)
        latest_result = None

        try:
            if len(voyage_stops) < 2:
                raise ValueError(
                    "Please add at least two ports."
                )

            speed = float(speed_input.value)
            restrictions = list(restriction_dropdown.value)

            routes = []

            for index in range(len(voyage_stops) - 1):

                route = calculate_route(
                    voyage_stops[index],
                    voyage_stops[index + 1],
                    restrictions
                )

                route["leg_number"] = index + 1

                routes.append(route)

            total_distance = sum(
                route["distance_nm"]
                for route in routes
            )

            duration = calculate_duration(
                total_distance,
                speed
            )

            latest_result = {
                "voyage_type": "multi_leg",
                "ports": list(voyage_stops),
                "speed_knots": speed,
                "restrictions": restrictions,
                "total_distance_nm": total_distance,
                "duration_hours": duration["hours"],
                "duration_days": duration["days"],
                "legs": routes
            }

            print("MULTI-LEG VOYAGE SUMMARY")
            print("-------------------------")

            results_df = pd.DataFrame([
                {
                    "Leg": route["leg_number"],
                    "Origin": route["origin"],
                    "Destination": route["destination"],
                    "Distance (NM)": round(
                        route["distance_nm"], 2
                    )
                }
                for route in routes
            ])

            display(results_df)

            print(
                f"\nTotal Distance: "
                f"{total_distance:,.2f} NM"
            )

            print(f"Vessel Speed: {speed:.1f} knots")

            print(
                "Restrictions:",
                ", ".join(restrictions) or "None"
            )

            print(
                f"Estimated Sailing Time: "
                f"{duration['hours']:.2f} hours"
            )

            print(
                f"Estimated Sailing Duration: "
                f"{duration['days']:.2f} days"
            )

            voyage_map = create_map(
                routes,
                voyage_stops
            )

            display(voyage_map)

        except Exception as error:
            print("Multi-leg calculation failed.")
            print("Error:", error)


calculate_multi_button.on_click(calculate_multi_clicked)


# ============================================================
# 9. ROUTE COMPARISON
# ============================================================

compare_button = widgets.Button(
    description="Compare Routes",
    button_style="info",
    icon="exchange"
)

comparison_output = widgets.Output()


def compare_routes_clicked(button):

    with comparison_output:

        clear_output(wait=True)

        try:
            origin_id = origin_dropdown.value
            destination_id = destination_dropdown.value

            if origin_id == destination_id:
                raise ValueError(
                    "Select two different ports."
                )

            selected_restrictions = list(
                restriction_dropdown.value
            )

            default_route = calculate_route(
                origin_id,
                destination_id,
                []
            )

            restricted_route = calculate_route(
                origin_id,
                destination_id,
                selected_restrictions
            )

            difference = (
                restricted_route["distance_nm"]
                - default_route["distance_nm"]
            )

            print("ROUTE COMPARISON")
            print("----------------")

            print(
                f"Default Route: "
                f"{default_route['distance_nm']:,.2f} NM"
            )

            print(
                f"Restricted Route: "
                f"{restricted_route['distance_nm']:,.2f} NM"
            )

            print(
                f"Distance Difference: "
                f"{difference:,.2f} NM"
            )

            comparison_map = folium.Map(
                location=[20, 40],
                zoom_start=2,
                tiles="OpenStreetMap"
            )

            all_coordinates = []

            for route, color, label in [
                (default_route, "blue", "Default"),
                (restricted_route, "red", "Restricted")
            ]:

                coordinates = [
                    [lat, lon]
                    for lon, lat in route["coordinates"]
                ]

                all_coordinates.extend(coordinates)

                folium.PolyLine(
                    locations=coordinates,
                    color=color,
                    weight=4,
                    opacity=0.85,
                    tooltip=(
                        f"{label}: "
                        f"{route['distance_nm']:,.2f} NM"
                    )
                ).add_to(comparison_map)

            for port_id, color in [
                (origin_id, "green"),
                (destination_id, "purple")
            ]:

                port = get_port(port_id)

                folium.Marker(
                    location=[
                        float(port["approach_latitude"]),
                        float(port["approach_longitude"])
                    ],
                    popup=port["port_name"],
                    icon=folium.Icon(color=color)
                ).add_to(comparison_map)

            comparison_map.fit_bounds(all_coordinates)

            legend_html = """
            <div style="
                position: fixed;
                bottom: 40px;
                left: 40px;
                width: 190px;
                background-color: white;
                border: 2px solid grey;
                z-index: 9999;
                padding: 10px;
                color: black;
                font-size: 13px;
            ">
                <b>Route Comparison</b><br>
                <span style="color:blue;">━━━</span>
                Default Route<br>
                <span style="color:red;">━━━</span>
                Restricted Route
            </div>
            """

            comparison_map.get_root().html.add_child(
                folium.Element(legend_html)
            )

            display(comparison_map)

        except Exception as error:
            print("Route comparison failed.")
            print("Error:", error)


compare_button.on_click(compare_routes_clicked)


# ============================================================
# 10. BUILD APPLICATION INTERFACE
# ============================================================

port_to_port_tab = widgets.VBox([
    widgets.HTML("<h3>Port-to-Port Voyage Planning</h3>"),
    origin_dropdown,
    destination_dropdown,
    port_route_button,
    port_output
])

multi_leg_tab = widgets.VBox([
    widgets.HTML("<h3>Multi-Leg Voyage Planning</h3>"),
    stop_dropdown,
    widgets.HBox([
        add_stop_button,
        remove_stop_button,
        clear_stops_button
    ]),
    itinerary_output,
    calculate_multi_button,
    multi_output
])

comparison_tab = widgets.VBox([
    widgets.HTML("<h3>Alternative Route Comparison</h3>"),
    widgets.HTML(
        "<p>Uses the origin and destination selected "
        "in the Port-to-Port tab and the shared "
        "passage restrictions.</p>"
    ),
    compare_button,
    comparison_output
])

tabs = widgets.Tab(
    children=[
        port_to_port_tab,
        multi_leg_tab,
        comparison_tab
    ]
)

tabs.set_title(0, "Port-to-Port")
tabs.set_title(1, "Multi-Leg")
tabs.set_title(2, "Route Comparison")


# ============================================================
# 11. DISPLAY APPLICATION
# ============================================================

display(
    widgets.HTML(
        """
        <h1>Maritime Voyage Planner</h1>
        <p>
        Project 6 — Maritime Distance and
        Voyage Calculation Service
        </p>
        <p>
        <b>Prototype:</b> Routes generated using
        the open-source searoute library.
        Distances and routes require validation
        before operational use.
        </p>
        """
    ),

    widgets.HTML("<h3>Voyage Settings</h3>"),

    speed_input,

    restriction_dropdown,

    widgets.HTML("<hr>"),

    tabs
)

display_itinerary()

print("\nMaritime Voyage Planner initialized successfully.")
```

```python
# ============================================================
# MARITIME VOYAGE PLANNER
# Project 6: Maritime Distance and Voyage Calculation Service
#
# Consolidated Jupyter Notebook Prototype
#
# Features:
# 1. Port-to-Port Voyage Planning and Comparison
# 2. Multi-Leg Voyage Planning and Comparison
# 3. Maritime Distance Calculation
# 4. Estimated Sailing Duration
# 5. Passage Restrictions
# 6. Interactive Maritime Maps
# 7. Input Validation
#
# Routing Engine: searoute (prototype)
# UI Framework: ipywidgets
# Map Framework: Folium
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import folium
import searoute as sr
import ipywidgets as widgets

from pathlib import Path
from IPython.display import display, clear_output


# ============================================================
# 2. LOAD PORT DATASET
# ============================================================

DATASET_PATH = Path("data/voyage_planner_datasets.xlsx")

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH.resolve()}"
    )

ports_df = pd.read_excel(
    DATASET_PATH,
    sheet_name="Port Reference"
)

active_ports = ports_df[
    ports_df["status"].astype(str).str.strip().str.lower() == "active"
].copy()

active_ports = active_ports[
    [
        "port_id",
        "port_name",
        "country",
        "latitude",
        "longitude",
        "approach_latitude",
        "approach_longitude"
    ]
]

active_ports = active_ports.sort_values(
    "port_name"
).reset_index(drop=True)

if active_ports.empty:
    raise ValueError("No active ports found in the dataset.")

port_lookup = active_ports.set_index("port_id")

port_options = [
    (
        f"{row['port_name']} ({row['country']})",
        row["port_id"]
    )
    for _, row in active_ports.iterrows()
]

print(f"Successfully loaded {len(active_ports)} active ports.")


# ============================================================
# 3. PASSAGE RESTRICTIONS
# ============================================================

PASSAGE_OPTIONS = [
    ("Automatic — No Restrictions", []),
    ("Avoid Suez Canal", ["suez"]),
    ("Avoid Panama Canal", ["panama"]),
    (
        "Avoid Suez and Panama Canals",
        ["suez", "panama"]
    ),
    (
        "Avoid Suez, Panama and Northwest Passage",
        ["suez", "panama", "northwest"]
    )
]


# ============================================================
# 4. HELPER FUNCTIONS
# ============================================================

def get_port(port_id):
    """Retrieve port information."""

    if port_id not in port_lookup.index:
        raise ValueError(f"Unknown port ID: {port_id}")

    return port_lookup.loc[port_id]


def get_coordinates(port_id):
    """Return coordinates in [longitude, latitude] format."""

    port = get_port(port_id)

    longitude = float(port["approach_longitude"])
    latitude = float(port["approach_latitude"])

    if not (-180 <= longitude <= 180):
        raise ValueError("Invalid port longitude.")

    if not (-90 <= latitude <= 90):
        raise ValueError("Invalid port latitude.")

    return [longitude, latitude]


def calculate_route(origin_id, destination_id, restrictions=None):
    """Calculate a maritime route using searoute."""

    if origin_id == destination_id:
        raise ValueError(
            "Origin and destination cannot be the same."
        )

    if restrictions is None:
        restrictions = []

    origin_coords = get_coordinates(origin_id)
    destination_coords = get_coordinates(destination_id)

    route = sr.searoute(
        origin_coords,
        destination_coords,
        units="naut",
        restrictions=restrictions
    )

    distance = float(route.properties["length"])
    coordinates = route.geometry["coordinates"]

    if not coordinates or distance <= 0:
        raise ValueError("No valid maritime route was returned.")

    return {
        "origin_id": origin_id,
        "destination_id": destination_id,
        "origin": get_port(origin_id)["port_name"],
        "destination": get_port(destination_id)["port_name"],
        "distance_nm": distance,
        "coordinates": coordinates,
        "restrictions": list(restrictions)
    }


def calculate_duration(distance_nm, speed_knots):
    """Calculate sailing duration."""

    if speed_knots <= 0:
        raise ValueError(
            "Vessel speed must be greater than zero."
        )

    hours = distance_nm / speed_knots

    return {
        "hours": hours,
        "days": hours / 24
    }


def calculate_itinerary_routes(port_ids, restrictions):
    """Calculate every leg in an ordered itinerary."""

    if len(port_ids) < 2:
        raise ValueError("Please select at least two ports.")

    routes = []

    for index in range(len(port_ids) - 1):

        route = calculate_route(
            port_ids[index],
            port_ids[index + 1],
            restrictions
        )

        route["leg_number"] = index + 1
        routes.append(route)

    return routes


def get_total_distance(routes):
    """Calculate the combined distance of all legs."""

    return sum(
        route["distance_nm"]
        for route in routes
    )


# ============================================================
# 5. MAP FUNCTIONS
# ============================================================

def add_port_markers(voyage_map, port_ids):
    """Add origin, intermediate and destination markers."""

    for index, port_id in enumerate(port_ids):

        port = get_port(port_id)

        if index == 0:
            marker_color = "green"

        elif index == len(port_ids) - 1:
            marker_color = "purple"

        else:
            marker_color = "blue"

        folium.Marker(
            location=[
                float(port["approach_latitude"]),
                float(port["approach_longitude"])
            ],
            popup=f"Stop {index + 1}: {port['port_name']}",
            tooltip=port["port_name"],
            icon=folium.Icon(
                color=marker_color,
                icon="info-sign"
            )
        ).add_to(voyage_map)


def create_map(routes, port_ids):
    """Create a standard maritime voyage map."""

    voyage_map = folium.Map(
        location=[20, 40],
        zoom_start=2,
        tiles="OpenStreetMap"
    )

    colors = [
        "blue",
        "green",
        "purple",
        "orange",
        "red",
        "darkblue"
    ]

    all_coordinates = []

    for index, route in enumerate(routes):

        coordinates = [
            [lat, lon]
            for lon, lat in route["coordinates"]
        ]

        all_coordinates.extend(coordinates)

        folium.PolyLine(
            locations=coordinates,
            color=colors[index % len(colors)],
            weight=4,
            opacity=0.85,
            tooltip=(
                f"Leg {index + 1}: "
                f"{route['origin']} → "
                f"{route['destination']} "
                f"({route['distance_nm']:,.2f} NM)"
            )
        ).add_to(voyage_map)

    add_port_markers(voyage_map, port_ids)

    if all_coordinates:
        voyage_map.fit_bounds(all_coordinates)

    return voyage_map


def create_comparison_map(
    default_routes,
    restricted_routes,
    port_ids
):
    """Display unrestricted and restricted routes together."""

    comparison_map = folium.Map(
        location=[20, 40],
        zoom_start=2,
        tiles="OpenStreetMap"
    )

    all_coordinates = []

    route_groups = [
        ("Unrestricted", default_routes, "blue"),
        ("Restricted", restricted_routes, "red")
    ]

    for label, routes, color in route_groups:

        for index, route in enumerate(routes):

            coordinates = [
                [lat, lon]
                for lon, lat in route["coordinates"]
            ]

            all_coordinates.extend(coordinates)

            folium.PolyLine(
                locations=coordinates,
                color=color,
                weight=4,
                opacity=0.85,
                tooltip=(
                    f"{label} Leg {index + 1}: "
                    f"{route['origin']} → "
                    f"{route['destination']} "
                    f"({route['distance_nm']:,.2f} NM)"
                )
            ).add_to(comparison_map)

    add_port_markers(comparison_map, port_ids)

    legend_html = """
    <div style="
        position: fixed;
        bottom: 40px;
        left: 40px;
        width: 220px;
        background-color: white;
        border: 2px solid grey;
        z-index: 9999;
        padding: 10px;
        color: black;
        font-size: 13px;
    ">
        <b>Route Comparison</b><br><br>
        <span style="color:blue;">━━━━</span>
        Unrestricted Route<br>
        <span style="color:red;">━━━━</span>
        Restricted Route<br><br>
        🟢 Origin<br>
        🔵 Intermediate Port<br>
        🟣 Final Destination
    </div>
    """

    comparison_map.get_root().html.add_child(
        folium.Element(legend_html)
    )

    if all_coordinates:
        comparison_map.fit_bounds(all_coordinates)

    return comparison_map


# ============================================================
# 6. SHARED APPLICATION STATE
# ============================================================

voyage_stops = []


# ============================================================
# 7. SHARED VOYAGE SETTINGS
# ============================================================

speed_input = widgets.BoundedFloatText(
    value=15.0,
    min=1.0,
    max=40.0,
    step=0.5,
    description="Vessel Speed (knots):",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="350px")
)

restriction_dropdown = widgets.Dropdown(
    options=PASSAGE_OPTIONS,
    description="Passage Restrictions:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="500px")
)


# ============================================================
# 8. PORT-TO-PORT VOYAGE PLANNER
# ============================================================

origin_dropdown = widgets.Dropdown(
    options=port_options,
    description="Origin:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

destination_dropdown = widgets.Dropdown(
    options=port_options,
    description="Destination:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

port_route_button = widgets.Button(
    description="Calculate Route",
    button_style="primary",
    icon="ship",
    layout=widgets.Layout(width="170px")
)

port_compare_button = widgets.Button(
    description="Compare Routes",
    button_style="info",
    icon="exchange",
    layout=widgets.Layout(width="170px")
)

port_output = widgets.Output()


def calculate_port_route_clicked(button):

    with port_output:

        clear_output(wait=True)

        try:
            origin_id = origin_dropdown.value
            destination_id = destination_dropdown.value

            speed = float(speed_input.value)
            restrictions = list(restriction_dropdown.value)

            route = calculate_route(
                origin_id,
                destination_id,
                restrictions
            )

            duration = calculate_duration(
                route["distance_nm"],
                speed
            )

            print("PORT-TO-PORT VOYAGE SUMMARY")
            print("=" * 55)

            print(f"Origin: {route['origin']}")
            print(f"Destination: {route['destination']}")

            print(
                "Passage Restrictions:",
                ", ".join(restrictions) or "None"
            )

            print(
                f"Maritime Distance: "
                f"{route['distance_nm']:,.2f} NM"
            )

            print(f"Vessel Speed: {speed:.1f} knots")

            print(
                f"Estimated Sailing Time: "
                f"{duration['hours']:.2f} hours"
            )

            print(
                f"Estimated Sailing Duration: "
                f"{duration['days']:.2f} days"
            )

            display(
                create_map(
                    [route],
                    [origin_id, destination_id]
                )
            )

        except Exception as error:
            print("Route calculation failed.")
            print("Error:", error)


def compare_port_routes_clicked(button):

    with port_output:

        clear_output(wait=True)

        try:
            origin_id = origin_dropdown.value
            destination_id = destination_dropdown.value

            speed = float(speed_input.value)
            restrictions = list(restriction_dropdown.value)

            default_route = calculate_route(
                origin_id,
                destination_id,
                []
            )

            restricted_route = calculate_route(
                origin_id,
                destination_id,
                restrictions
            )

            default_distance = default_route["distance_nm"]
            restricted_distance = restricted_route["distance_nm"]

            difference = restricted_distance - default_distance

            default_duration = calculate_duration(
                default_distance,
                speed
            )

            restricted_duration = calculate_duration(
                restricted_distance,
                speed
            )

            print("PORT-TO-PORT ROUTE COMPARISON")
            print("=" * 55)

            print(
                f"Origin: {default_route['origin']}"
            )

            print(
                f"Destination: {default_route['destination']}"
            )

            print(
                "Passage Restrictions:",
                ", ".join(restrictions) or "None"
            )

            comparison_df = pd.DataFrame({
                "Metric": [
                    "Distance (NM)",
                    "Sailing Time (Hours)",
                    "Sailing Duration (Days)"
                ],
                "Unrestricted Route": [
                    round(default_distance, 2),
                    round(default_duration["hours"], 2),
                    round(default_duration["days"], 2)
                ],
                "Restricted Route": [
                    round(restricted_distance, 2),
                    round(restricted_duration["hours"], 2),
                    round(restricted_duration["days"], 2)
                ],
                "Difference": [
                    round(difference, 2),
                    round(
                        restricted_duration["hours"]
                        - default_duration["hours"],
                        2
                    ),
                    round(
                        restricted_duration["days"]
                        - default_duration["days"],
                        2
                    )
                ]
            })

            display(comparison_df)

            display(
                create_comparison_map(
                    [default_route],
                    [restricted_route],
                    [origin_id, destination_id]
                )
            )

        except Exception as error:
            print("Port-to-port comparison failed.")
            print("Error:", error)


port_route_button.on_click(calculate_port_route_clicked)
port_compare_button.on_click(compare_port_routes_clicked)


# ============================================================
# 9. MULTI-LEG VOYAGE PLANNER
# ============================================================

stop_dropdown = widgets.Dropdown(
    options=port_options,
    description="Select Port:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

add_stop_button = widgets.Button(
    description="Add Stop",
    button_style="success",
    icon="plus"
)

remove_stop_button = widgets.Button(
    description="Remove Last Stop",
    button_style="warning",
    icon="minus"
)

clear_stops_button = widgets.Button(
    description="Clear All",
    button_style="danger",
    icon="trash"
)

calculate_multi_button = widgets.Button(
    description="Calculate Multi-Leg Voyage",
    button_style="primary",
    icon="ship",
    layout=widgets.Layout(width="250px")
)

compare_multi_button = widgets.Button(
    description="Compare Multi-Leg Routes",
    button_style="info",
    icon="exchange",
    layout=widgets.Layout(width="250px")
)

itinerary_output = widgets.Output()
multi_output = widgets.Output()


def display_itinerary():

    with itinerary_output:

        clear_output(wait=True)

        print("VOYAGE ITINERARY")
        print("-" * 35)

        if not voyage_stops:
            print("No ports selected.")
            return

        for index, port_id in enumerate(voyage_stops, 1):

            port = get_port(port_id)

            print(
                f"{index}. {port['port_name']}"
            )


def add_stop_clicked(button):

    selected_port = stop_dropdown.value

    if voyage_stops and voyage_stops[-1] == selected_port:

        with itinerary_output:

            clear_output(wait=True)

            print(
                "Error: Consecutive stops "
                "cannot be the same port."
            )

        return

    voyage_stops.append(selected_port)

    display_itinerary()


def remove_stop_clicked(button):

    if voyage_stops:
        voyage_stops.pop()

    display_itinerary()


def clear_stops_clicked(button):

    voyage_stops.clear()

    display_itinerary()


add_stop_button.on_click(add_stop_clicked)
remove_stop_button.on_click(remove_stop_clicked)
clear_stops_button.on_click(clear_stops_clicked)


def calculate_multi_clicked(button):

    with multi_output:

        clear_output(wait=True)

        try:
            speed = float(speed_input.value)
            restrictions = list(restriction_dropdown.value)

            routes = calculate_itinerary_routes(
                voyage_stops,
                restrictions
            )

            total_distance = get_total_distance(routes)

            duration = calculate_duration(
                total_distance,
                speed
            )

            print("MULTI-LEG VOYAGE SUMMARY")
            print("=" * 60)

            results_df = pd.DataFrame([
                {
                    "Leg": route["leg_number"],
                    "Origin": route["origin"],
                    "Destination": route["destination"],
                    "Distance (NM)": round(
                        route["distance_nm"], 2
                    )
                }
                for route in routes
            ])

            display(results_df)

            print(
                f"\nTotal Distance: "
                f"{total_distance:,.2f} NM"
            )

            print(f"Vessel Speed: {speed:.1f} knots")

            print(
                "Passage Restrictions:",
                ", ".join(restrictions) or "None"
            )

            print(
                f"Estimated Sailing Time: "
                f"{duration['hours']:.2f} hours"
            )

            print(
                f"Estimated Sailing Duration: "
                f"{duration['days']:.2f} days"
            )

            display(
                create_map(
                    routes,
                    voyage_stops
                )
            )

        except Exception as error:
            print("Multi-leg calculation failed.")
            print("Error:", error)


def compare_multi_routes_clicked(button):

    with multi_output:

        clear_output(wait=True)

        try:
            speed = float(speed_input.value)
            restrictions = list(restriction_dropdown.value)

            default_routes = calculate_itinerary_routes(
                voyage_stops,
                []
            )

            restricted_routes = calculate_itinerary_routes(
                voyage_stops,
                restrictions
            )

            comparison_rows = []

            for index, (default_route, restricted_route) in enumerate(
                zip(default_routes, restricted_routes),
                start=1
            ):

                default_distance = default_route["distance_nm"]
                restricted_distance = restricted_route["distance_nm"]

                comparison_rows.append({
                    "Leg": index,
                    "Origin": default_route["origin"],
                    "Destination": default_route["destination"],
                    "Unrestricted (NM)": round(
                        default_distance, 2
                    ),
                    "Restricted (NM)": round(
                        restricted_distance, 2
                    ),
                    "Difference (NM)": round(
                        restricted_distance - default_distance,
                        2
                    )
                })

            total_default = get_total_distance(default_routes)
            total_restricted = get_total_distance(restricted_routes)

            default_duration = calculate_duration(
                total_default,
                speed
            )

            restricted_duration = calculate_duration(
                total_restricted,
                speed
            )

            print("MULTI-LEG ROUTE COMPARISON")
            print("=" * 60)

            print(
                "Passage Restrictions:",
                ", ".join(restrictions) or "None"
            )

            print(f"Vessel Speed: {speed:.1f} knots")

            print("\nINDIVIDUAL LEG COMPARISON")

            display(
                pd.DataFrame(comparison_rows)
            )

            print("\nTOTAL VOYAGE COMPARISON")

            summary_df = pd.DataFrame({
                "Metric": [
                    "Total Distance (NM)",
                    "Sailing Time (Hours)",
                    "Sailing Duration (Days)"
                ],
                "Unrestricted Route": [
                    round(total_default, 2),
                    round(default_duration["hours"], 2),
                    round(default_duration["days"], 2)
                ],
                "Restricted Route": [
                    round(total_restricted, 2),
                    round(restricted_duration["hours"], 2),
                    round(restricted_duration["days"], 2)
                ],
                "Difference": [
                    round(
                        total_restricted - total_default,
                        2
                    ),
                    round(
                        restricted_duration["hours"]
                        - default_duration["hours"],
                        2
                    ),
                    round(
                        restricted_duration["days"]
                        - default_duration["days"],
                        2
                    )
                ]
            })

            display(summary_df)

            display(
                create_comparison_map(
                    default_routes,
                    restricted_routes,
                    voyage_stops
                )
            )

        except Exception as error:
            print("Multi-leg comparison failed.")
            print("Error:", error)


calculate_multi_button.on_click(calculate_multi_clicked)
compare_multi_button.on_click(compare_multi_routes_clicked)


# ============================================================
# 10. BUILD APPLICATION INTERFACE
# ============================================================

port_to_port_tab = widgets.VBox([
    widgets.HTML("<h3>Port-to-Port Voyage Planning</h3>"),
    origin_dropdown,
    destination_dropdown,
    widgets.HBox([
        port_route_button,
        port_compare_button
    ]),
    port_output
])

multi_leg_tab = widgets.VBox([
    widgets.HTML("<h3>Multi-Leg Voyage Planning</h3>"),
    stop_dropdown,
    widgets.HBox([
        add_stop_button,
        remove_stop_button,
        clear_stops_button
    ]),
    itinerary_output,
    widgets.HBox([
        calculate_multi_button,
        compare_multi_button
    ]),
    multi_output
])

tabs = widgets.Tab(
    children=[
        port_to_port_tab,
        multi_leg_tab
    ]
)

tabs.set_title(0, "Port-to-Port")
tabs.set_title(1, "Multi-Leg")


# ============================================================
# 11. DISPLAY APPLICATION
# ============================================================

display(
    widgets.HTML(
        """
        <h1>Maritime Voyage Planner</h1>
        <p>
            Project 6 — Maritime Distance and
            Voyage Calculation Service
        </p>
        <p>
            <b>Prototype:</b> Routes generated using
            the open-source searoute library.
            Distances and routes require validation
            before operational use.
        </p>
        """
    ),

    widgets.HTML("<h3>Voyage Settings</h3>"),

    speed_input,

    restriction_dropdown,

    widgets.HTML("<hr>"),

    tabs
)

display_itinerary()

print("\nMaritime Voyage Planner initialized successfully.")
```

```python
# ============================================================
# POSITION-TO-PORT VOYAGE PLANNER
# STEP 1: CREATE INPUT INTERFACE
# ============================================================

# Vessel latitude
position_latitude = widgets.BoundedFloatText(
    value=1.2903,
    min=-90,
    max=90,
    step=0.0001,
    description="Vessel Latitude:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="350px")
)

# Vessel longitude
position_longitude = widgets.BoundedFloatText(
    value=103.8519,
    min=-180,
    max=180,
    step=0.0001,
    description="Vessel Longitude:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="350px")
)

# Destination port
position_destination = widgets.Dropdown(
    options=port_options,
    description="Destination Port:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

# Calculate route button
position_route_button = widgets.Button(
    description="Calculate Position-to-Port Route",
    button_style="primary",
    icon="ship",
    layout=widgets.Layout(width="280px")
)

# Output area
position_output = widgets.Output()

# Display interface
display(
    widgets.HTML("<h3>Position-to-Port Voyage Planning</h3>"),
    position_latitude,
    position_longitude,
    position_destination,
    position_route_button,
    position_output
)
```

```python
# ============================================================
# POSITION-TO-PORT VOYAGE PLANNER
# STEP 2: ROUTE CALCULATION AND VISUALIZATION
# ============================================================

def calculate_position_to_port(button):

    with position_output:

        clear_output(wait=True)

        try:
            # Retrieve inputs
            latitude = float(position_latitude.value)
            longitude = float(position_longitude.value)

            destination_id = position_destination.value
            destination = get_port(destination_id)

            speed = float(speed_input.value)
            restrictions = list(restriction_dropdown.value)

            # Validate inputs
            if not (-90 <= latitude <= 90):
                raise ValueError("Latitude must be between -90 and 90.")

            if not (-180 <= longitude <= 180):
                raise ValueError("Longitude must be between -180 and 180.")

            if speed <= 0:
                raise ValueError("Vessel speed must be greater than zero.")

            # Coordinates must be [longitude, latitude]
            vessel_coordinates = [longitude, latitude]
            destination_coordinates = get_coordinates(destination_id)

            # Calculate route
            route = sr.searoute(
                vessel_coordinates,
                destination_coordinates,
                units="naut",
                restrictions=restrictions
            )

            distance = float(route.properties["length"])

            if distance <= 0:
                raise ValueError("No valid maritime route was returned.")

            duration = calculate_duration(distance, speed)

            # Display results
            print("POSITION-TO-PORT VOYAGE SUMMARY")
            print("=" * 50)

            print(f"Vessel Position: {latitude:.4f}, {longitude:.4f}")
            print(f"Destination: {destination['port_name']}")
            print(f"Maritime Distance: {distance:,.2f} NM")
            print(f"Vessel Speed: {speed:.1f} knots")
            print(f"Estimated Sailing Time: {duration['hours']:.2f} hours")
            print(f"Estimated Sailing Duration: {duration['days']:.2f} days")
            print(f"Passage Restrictions: {', '.join(restrictions) or 'None'}")

            # Create map
            voyage_map = folium.Map(
                location=[latitude, longitude],
                zoom_start=3
            )

            route_coordinates = [
                [lat, lon]
                for lon, lat in route.geometry["coordinates"]
            ]

            # Draw route
            folium.PolyLine(
                locations=route_coordinates,
                color="blue",
                weight=4,
                tooltip=f"{distance:,.2f} NM"
            ).add_to(voyage_map)

            # Vessel marker
            folium.Marker(
                location=[latitude, longitude],
                popup="Current Vessel Position",
                tooltip="Vessel Position",
                icon=folium.Icon(color="green")
            ).add_to(voyage_map)

            # Destination marker
            folium.Marker(
                location=[
                    float(destination["approach_latitude"]),
                    float(destination["approach_longitude"])
                ],
                popup=destination["port_name"],
                tooltip="Destination Port",
                icon=folium.Icon(color="purple")
            ).add_to(voyage_map)

            # Fit map
            voyage_map.fit_bounds(
                route_coordinates + [
                    [latitude, longitude],
                    [
                        float(destination["approach_latitude"]),
                        float(destination["approach_longitude"])
                    ]
                ]
            )

            display(voyage_map)

        except Exception as error:
            print("Position-to-port calculation failed.")
            print("Error:", error)


# Connect button to calculation
position_route_button.on_click(calculate_position_to_port)

print("Position-to-Port routing engine connected successfully.")
```

```python
# ============================================================
# POSITION-TO-PORT VOYAGE PLANNER
# STEP 4: ROUTE COMPARISON
# ============================================================

# 1. CREATE COMPARE ROUTES BUTTON

position_compare_button = widgets.Button(
    description="Compare Routes",
    button_style="info",
    icon="exchange",
    layout=widgets.Layout(width="180px")
)


# 2. CREATE ROUTE COMPARISON FUNCTION

def compare_position_to_port(button):

    with position_output:

        clear_output(wait=True)

        try:

            # Retrieve vessel coordinates
            latitude = float(position_latitude.value)
            longitude = float(position_longitude.value)

            # Retrieve destination
            destination_id = position_destination.value
            destination = get_port(destination_id)

            # Retrieve voyage settings
            speed = float(speed_input.value)
            restrictions = list(restriction_dropdown.value)

            # Validate inputs
            if not (-90 <= latitude <= 90):
                raise ValueError(
                    "Latitude must be between -90 and 90."
                )

            if not (-180 <= longitude <= 180):
                raise ValueError(
                    "Longitude must be between -180 and 180."
                )

            if speed <= 0:
                raise ValueError(
                    "Vessel speed must be greater than zero."
                )

            # Validate passage restriction selection
            if not restrictions:
                raise ValueError(
                    "Please select a passage restriction "
                    "before comparing routes."
                )

            # searoute requires [longitude, latitude]
            vessel_coordinates = [longitude, latitude]

            destination_coordinates = get_coordinates(
                destination_id
            )

            # ------------------------------------------------
            # CALCULATE UNRESTRICTED ROUTE
            # ------------------------------------------------

            unrestricted_route = sr.searoute(
                vessel_coordinates,
                destination_coordinates,
                units="naut",
                restrictions=[]
            )

            unrestricted_distance = float(
                unrestricted_route.properties["length"]
            )

            # ------------------------------------------------
            # CALCULATE RESTRICTED ROUTE
            # ------------------------------------------------

            restricted_route = sr.searoute(
                vessel_coordinates,
                destination_coordinates,
                units="naut",
                restrictions=restrictions
            )

            restricted_distance = float(
                restricted_route.properties["length"]
            )

            if unrestricted_distance <= 0 or restricted_distance <= 0:
                raise ValueError(
                    "One or both routes returned an invalid distance."
                )

            # ------------------------------------------------
            # CALCULATE SAILING DURATIONS
            # ------------------------------------------------

            unrestricted_duration = calculate_duration(
                unrestricted_distance,
                speed
            )

            restricted_duration = calculate_duration(
                restricted_distance,
                speed
            )

            distance_difference = (
                restricted_distance - unrestricted_distance
            )

            duration_difference = (
                restricted_duration["days"]
                - unrestricted_duration["days"]
            )

            # ------------------------------------------------
            # DISPLAY COMPARISON SUMMARY
            # ------------------------------------------------

            print("POSITION-TO-PORT ROUTE COMPARISON")
            print("=" * 60)

            print(
                f"Vessel Position: "
                f"{latitude:.4f}, {longitude:.4f}"
            )

            print(
                f"Destination: {destination['port_name']}"
            )

            print(f"Vessel Speed: {speed:.1f} knots")

            print(
                f"Selected Restrictions: "
                f"{', '.join(restrictions)}"
            )

            comparison_df = pd.DataFrame([
                {
                    "Route": "Unrestricted",
                    "Distance (NM)": round(
                        unrestricted_distance, 2
                    ),
                    "Duration (Days)": round(
                        unrestricted_duration["days"], 2
                    )
                },
                {
                    "Route": "Restricted",
                    "Distance (NM)": round(
                        restricted_distance, 2
                    ),
                    "Duration (Days)": round(
                        restricted_duration["days"], 2
                    )
                },
                {
                    "Route": "Difference",
                    "Distance (NM)": round(
                        distance_difference, 2
                    ),
                    "Duration (Days)": round(
                        duration_difference, 2
                    )
                }
            ])

            display(comparison_df)

            # ------------------------------------------------
            # CREATE COMPARISON MAP
            # ------------------------------------------------

            comparison_map = folium.Map(
                location=[latitude, longitude],
                zoom_start=3,
                tiles="OpenStreetMap"
            )

            # Convert GeoJSON coordinates
            unrestricted_coordinates = [
                [lat, lon]
                for lon, lat in unrestricted_route.geometry["coordinates"]
            ]

            restricted_coordinates = [
                [lat, lon]
                for lon, lat in restricted_route.geometry["coordinates"]
            ]

            # Unrestricted route — SOLID BLUE
            folium.PolyLine(
                locations=unrestricted_coordinates,
                color="blue",
                weight=4,
                opacity=0.85,
                tooltip=(
                    f"Unrestricted Route: "
                    f"{unrestricted_distance:,.2f} NM"
                )
            ).add_to(comparison_map)

            # Restricted route — SOLID RED
            folium.PolyLine(
                locations=restricted_coordinates,
                color="red",
                weight=4,
                opacity=0.85,
                tooltip=(
                    f"Restricted Route: "
                    f"{restricted_distance:,.2f} NM"
                )
            ).add_to(comparison_map)

            # ------------------------------------------------
            # ADD VESSEL AND DESTINATION MARKERS
            # ------------------------------------------------

            folium.Marker(
                location=[latitude, longitude],
                popup="Current Vessel Position",
                tooltip="Vessel Position",
                icon=folium.Icon(color="green")
            ).add_to(comparison_map)

            destination_location = [
                float(destination["approach_latitude"]),
                float(destination["approach_longitude"])
            ]

            folium.Marker(
                location=destination_location,
                popup=destination["port_name"],
                tooltip="Destination Port",
                icon=folium.Icon(color="purple")
            ).add_to(comparison_map)

            # ------------------------------------------------
            # ADD MAP LEGEND
            # ------------------------------------------------

            legend_html = """
            <div style="
                position: fixed;
                bottom: 30px;
                left: 30px;
                z-index: 9999;
                background-color: white;
                border: 2px solid grey;
                border-radius: 6px;
                padding: 12px;
                font-size: 13px;
                color: black;
            ">
                <b>Route Comparison</b><br><br>

                <span style="color: blue;">
                    ━━━━━
                </span>
                Unrestricted Route<br>

                <span style="color: red;">
                    ━━━━━
                </span>
                Restricted Route
            </div>
            """

            comparison_map.get_root().html.add_child(
                folium.Element(legend_html)
            )

            # ------------------------------------------------
            # FIT MAP TO BOTH ROUTES
            # ------------------------------------------------

            all_coordinates = (
                unrestricted_coordinates
                + restricted_coordinates
                + [[latitude, longitude]]
                + [destination_location]
            )

            comparison_map.fit_bounds(all_coordinates)

            display(comparison_map)

        except Exception as error:

            print("Position-to-port route comparison failed.")
            print("Error:", error)


# 3. CONNECT COMPARE BUTTON

position_compare_button.on_click(
    compare_position_to_port
)


# 4. DISPLAY BOTH BUTTONS TOGETHER

display(
    widgets.HTML(
        "<h3>Position-to-Port Voyage Planning</h3>"
    ),

    position_latitude,

    position_longitude,

    position_destination,

    widgets.HBox([
        position_route_button,
        position_compare_button
    ]),

    position_output
)

print("Position-to-Port route comparison connected successfully.")
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import folium
import searoute as sr

from pathlib import Path
from streamlit.components.v1 import html


# ============================================================
# APPLICATION CONFIGURATION
# ============================================================

st.set_page_config(
    page_title="Maritime Voyage Planner",
    page_icon="🚢",
    layout="wide"
)

st.title("🚢 Maritime Voyage Planner")


# ============================================================
# LOAD PORT DATASET
# ============================================================

DATASET_PATH = Path("data/voyage_planner_datasets.xlsx")

@st.cache_data
def load_ports():

    df = pd.read_excel(
        DATASET_PATH,
        sheet_name="Port Reference"
    )

    df = df[
        df["status"].astype(str).str.lower() == "active"
    ].copy()

    df = df.sort_values("port_name")

    return df


try:
    ports_df = load_ports()

except Exception as error:
    st.error(f"Failed to load port dataset: {error}")
    st.stop()


port_options = {
    f"{row['port_name']} ({row['country']})": row
    for _, row in ports_df.iterrows()
}

port_names = list(port_options.keys())


# ============================================================
# VOYAGE SETTINGS
# ============================================================

st.subheader("Voyage Settings")

speed = st.number_input(
    "Vessel Speed (knots)",
    min_value=0.1,
    max_value=50.0,
    value=15.0,
    step=0.5
)

restriction_options = {
    "Automatic — No Restrictions": [],
    "Avoid Suez Canal": ["suez"],
    "Avoid Panama Canal": ["panama"],
    "Avoid Suez and Panama Canals": ["suez", "panama"],
    "Avoid Suez, Panama and Northwest Passage": [
        "suez", "panama", "northwest"
    ]
}

restriction_selection = st.selectbox(
    "Passage Restrictions",
    list(restriction_options.keys())
)

restrictions = restriction_options[restriction_selection]


# ============================================================
# ROUTING FUNCTIONS
# ============================================================

def get_coordinates(port):

    return [
        float(port["approach_longitude"]),
        float(port["approach_latitude"])
    ]


def calculate_route(origin, destination):

    route = sr.searoute(
        get_coordinates(origin),
        get_coordinates(destination),
        units="naut",
        restrictions=restrictions
    )

    distance = float(route.properties["length"])

    duration_hours = distance / speed
    duration_days = duration_hours / 24

    return route, distance, duration_hours, duration_days


def create_route_map(route, origin, destination):

    coordinates = [
        [lat, lon]
        for lon, lat in route.geometry["coordinates"]
    ]

    voyage_map = folium.Map(
        location=coordinates[0],
        zoom_start=3
    )

    folium.PolyLine(
        coordinates,
        color="blue",
        weight=4,
        tooltip="Maritime Route"
    ).add_to(voyage_map)

    folium.Marker(
        [
            float(origin["approach_latitude"]),
            float(origin["approach_longitude"])
        ],
        tooltip="Origin",
        popup=origin["port_name"],
        icon=folium.Icon(color="green")
    ).add_to(voyage_map)

    folium.Marker(
        [
            float(destination["approach_latitude"]),
            float(destination["approach_longitude"])
        ],
        tooltip="Destination",
        popup=destination["port_name"],
        icon=folium.Icon(color="purple")
    ).add_to(voyage_map)

    voyage_map.fit_bounds(coordinates)

    return voyage_map


# ============================================================
# APPLICATION TABS
# ============================================================

tab1, tab2, tab3 = st.tabs([
    "Port-to-Port",
    "Multi-Leg",
    "Position-to-Port"
])


# ============================================================
# TAB 1 — PORT-TO-PORT
# ============================================================

with tab1:

    st.header("Port-to-Port Voyage Planning")

    origin_name = st.selectbox(
        "Origin Port",
        port_names,
        key="origin"
    )

    destination_name = st.selectbox(
        "Destination Port",
        port_names,
        key="destination",
        index=min(1, len(port_names) - 1)
    )

    if st.button("Calculate Route", key="calculate_port"):

        try:

            if origin_name == destination_name:
                st.error(
                    "Origin and destination ports must be different."
                )

            else:

                origin = port_options[origin_name]
                destination = port_options[destination_name]

                route, distance, hours, days = calculate_route(
                    origin,
                    destination
                )

                st.success("Route calculated successfully!")

                col1, col2, col3 = st.columns(3)

                col1.metric(
                    "Maritime Distance",
                    f"{distance:,.2f} NM"
                )

                col2.metric(
                    "Sailing Time",
                    f"{hours:,.2f} hours"
                )

                col3.metric(
                    "Sailing Duration",
                    f"{days:,.2f} days"
                )

                voyage_map = create_route_map(
                    route,
                    origin,
                    destination
                )

                html(
                    voyage_map._repr_html_(),
                    height=550
                )

        except Exception as error:

            st.error(f"Route calculation failed: {error}")


# ============================================================
# TAB 2 — MULTI-LEG
# ============================================================

with tab2:

    st.header("Multi-Leg Voyage Planning")

    st.info(
        "Multi-Leg functionality will be migrated next."
    )


# ============================================================
# TAB 3 — POSITION-TO-PORT
# ============================================================

with tab3:

    st.header("Position-to-Port Voyage Planning")

    st.info(
        "Position-to-Port functionality will be migrated after Multi-Leg."
    )
```

```python
from pathlib import Path
import shutil

# ============================================================
# STEP 8A — SHARED INPUT VALIDATION
# ============================================================

app_path = Path("app.py")
app_code = app_path.read_text(encoding="utf-8")

validation_marker = "# VALIDATION FUNCTIONS — STEP 8A"

if validation_marker in app_code:
    raise ValueError(
        "Validation functions are already installed."
    )

validation_functions = '''
# ============================================================
# VALIDATION FUNCTIONS — STEP 8A
# ============================================================

import math


def validate_coordinates(longitude, latitude):
    """Validate longitude and latitude values."""

    try:
        longitude = float(longitude)
        latitude = float(latitude)
    except (TypeError, ValueError):
        raise ValueError(
            "Longitude and latitude must be numeric values."
        )

    if not math.isfinite(longitude) or not math.isfinite(latitude):
        raise ValueError(
            "Longitude and latitude must be finite numbers."
        )

    if not -180 <= longitude <= 180:
        raise ValueError(
            "Longitude must be between -180 and 180 degrees."
        )

    if not -90 <= latitude <= 90:
        raise ValueError(
            "Latitude must be between -90 and 90 degrees."
        )

    return [longitude, latitude]


def validate_speed(vessel_speed):
    """Ensure vessel speed is valid."""

    try:
        vessel_speed = float(vessel_speed)
    except (TypeError, ValueError):
        raise ValueError(
            "Vessel speed must be a numeric value."
        )

    if not math.isfinite(vessel_speed) or vessel_speed <= 0:
        raise ValueError(
            "Vessel speed must be greater than zero."
        )

    return vessel_speed


def validate_route_result(route):
    """Check that the routing engine returned a usable route."""

    if route is None:
        raise ValueError(
            "The routing engine did not return a route."
        )

    if not hasattr(route, "properties"):
        raise ValueError(
            "The routing engine returned an invalid route."
        )

    distance = float(route.properties["length"])

    if not math.isfinite(distance) or distance <= 0:
        raise ValueError(
            "The calculated route distance must be greater than zero."
        )

    geometry = route.geometry

    if not geometry or not geometry.get("coordinates"):
        raise ValueError(
            "The routing engine returned no route coordinates."
        )

    coordinates = geometry["coordinates"]

    if len(coordinates) < 2:
        raise ValueError(
            "The calculated route contains insufficient coordinates."
        )

    for coordinate in coordinates:
        if len(coordinate) < 2:
            raise ValueError(
                "The route contains an invalid coordinate."
            )

        validate_coordinates(
            coordinate[0],
            coordinate[1]
        )

    return distance


def validate_itinerary(stops):
    """Validate a multi-leg voyage itinerary."""

    if len(stops) < 2:
        raise ValueError(
            "A multi-leg voyage requires at least two ports."
        )

    for index in range(len(stops) - 1):
        if stops[index] == stops[index + 1]:
            raise ValueError(
                "Consecutive duplicate ports are not allowed."
            )

    return True
'''

# Insert before the application tabs
tabs_marker = "# APPLICATION TABS"

if tabs_marker not in app_code:
    raise ValueError(
        "Application tabs marker not found. "
        "No changes were made."
    )

marker_index = app_code.index(tabs_marker)

updated_code = (
    app_code[:marker_index]
    + validation_functions
    + "\n"
    + app_code[marker_index:]
)

# Check syntax before saving
compile(updated_code, "app.py", "exec")

# Create backup
backup_path = Path("app_before_validation.py")
shutil.copy2(app_path, backup_path)

# Save updated application
app_path.write_text(
    updated_code,
    encoding="utf-8"
)

print("SUCCESS: Validation functions installed!")
print("Backup saved as:", backup_path)
```

```python
from pathlib import Path
import shutil
import re

# ============================================================
# STEP 8C.2 — ERROR RECOVERY
# ============================================================

app_path = Path("app.py")

if not app_path.exists():
    raise FileNotFoundError("app.py not found.")

original_code = app_path.read_text(encoding="utf-8")

marker = "# ERROR RECOVERY — STEP 8C.2"

if marker in original_code:
    raise ValueError("Error recovery is already installed.")

# Map each error handler to its persistence key.
error_handlers = {
    "Route calculation failed:": "persist_calculate_port",
    "Route comparison failed:": "persist_compare_port",
    "Multi-leg calculation failed:": "persist_calculate_multi",
    "Multi-leg comparison failed:": "persist_compare_multi",
    "Position-to-Port calculation failed:": "persist_calculate_position",
    "Position-to-Port comparison failed:": "persist_compare_position"
}

updated_code = original_code

for error_message, state_key in error_handlers.items():

    # Find the exception handler associated with this message.
    pattern = (
        r"(?P<indent>^[ \t]*)except Exception as error:"
        r"(?P<body>[\s\S]*?)"
        + re.escape(error_message)
    )

    matches = list(
        re.finditer(
            pattern,
            updated_code,
            flags=re.MULTILINE
        )
    )

    if len(matches) != 1:
        raise ValueError(
            f"Could not uniquely identify handler: {error_message}. "
            "No changes were saved."
        )

    match = matches[0]
    indentation = match.group("indent")

    replacement = (
        f"{indentation}except Exception as error:\n"
        f"{indentation}    {marker}\n"
        f"{indentation}    st.session_state.pop("
        f'"{state_key}", None)\n'
    )

    updated_code = (
        updated_code[:match.start()]
        + replacement
        + updated_code[match.end() - len(match.group("body")):]
    )

# Validate Python syntax.
compile(updated_code, "app.py", "exec")

# Create backup.
backup_path = Path("app_before_error_recovery.py")
shutil.copy2(app_path, backup_path)

# Save updated application.
app_path.write_text(
    updated_code,
    encoding="utf-8"
)

print("SUCCESS: Error recovery installed!")
print("Backup saved as:", backup_path)
```

```python
from pathlib import Path
import shutil
import re

# ============================================================
# STEP 8C.2 — SAFE ERROR RECOVERY INSTALLATION
# ============================================================

app_path = Path("app.py")
original_code = app_path.read_text(encoding="utf-8")

marker = "# ERROR RECOVERY — STEP 8C.2"

if marker in original_code:
    raise ValueError("Error recovery is already installed.")

# Each error message corresponds to one routing operation.
error_handlers = {
    "Route calculation failed:": "persist_calculate_port",
    "Route comparison failed:": "persist_compare_port",
    "Multi-leg calculation failed:": "persist_calculate_multi",
    "Multi-leg comparison failed:": "persist_compare_multi",
    "Position-to-Port calculation failed:": "persist_calculate_position",
    "Position-to-Port comparison failed:": "persist_compare_position"
}

lines = original_code.splitlines(keepends=True)
insertions = []

# ------------------------------------------------------------
# 1. IDENTIFY EXACT EXCEPTION HANDLERS
# ------------------------------------------------------------

for error_message, state_key in error_handlers.items():

    matches = [
        i for i, line in enumerate(lines)
        if error_message in line
    ]

    if len(matches) != 1:
        raise ValueError(
            f"Expected one occurrence of '{error_message}', "
            f"found {len(matches)}. No changes were made."
        )

    error_line_index = matches[0]

    # Search backward for the nearest exception handler.
    except_index = None

    for i in range(error_line_index - 1, -1, -1):

        if re.match(
            r"^\s*except Exception as error:\s*$",
            lines[i]
        ):
            except_index = i
            break

    if except_index is None:
        raise ValueError(
            f"Exception handler not found for {error_message}."
        )

    # Preserve the original indentation.
    except_line = lines[except_index]

    indentation = (
        except_line[:len(except_line) - len(except_line.lstrip())]
    )

    insertion = (
        f"{indentation}    {marker}\n"
        f'{indentation}    st.session_state.pop("{state_key}", None)\n'
    )

    insertions.append((except_index + 1, insertion))

# Ensure each error message maps to a different exception handler.
positions = [position for position, _ in insertions]

if len(positions) != len(set(positions)):
    raise ValueError(
        "Multiple error messages matched the same exception handler. "
        "No changes were made."
    )

# ------------------------------------------------------------
# 2. INSERT ERROR RECOVERY STATEMENTS
# ------------------------------------------------------------

# Insert from bottom to top so line positions remain valid.
for position, insertion in sorted(
    insertions,
    reverse=True
):
    lines.insert(position, insertion)

updated_code = "".join(lines)

# ------------------------------------------------------------
# 3. VALIDATE BEFORE SAVING
# ------------------------------------------------------------

compile(updated_code, "app.py", "exec")

# Verify all six persistence keys were inserted.
for state_key in error_handlers.values():

    expected = (
        f'st.session_state.pop("{state_key}", None)'
    )

    if expected not in updated_code:
        raise ValueError(
            f"Missing error recovery for {state_key}."
        )

# ------------------------------------------------------------
# 4. BACKUP AND SAVE
# ------------------------------------------------------------

backup_path = Path("app_before_error_recovery.py")

shutil.copy2(app_path, backup_path)

app_path.write_text(
    updated_code,
    encoding="utf-8"
)

print("SUCCESS: Error recovery installed!")
print("Backup saved as:", backup_path)
print("All six routing error handlers updated.")
```
